```python
import torch

def f(x, y):
  # трьохгорбна функція
  return 2*x**2  - 1.05*x**4 + 1/6*x**6 + x*y  + y**2


result_true = 0
```

```python
x = torch.tensor(1.2, device='cuda',  requires_grad=True)
y = torch.tensor(1.2, device='cuda',  requires_grad=True)
print(x)
y
```

```python
optimizer = torch.optim.SGD([x, y], lr=1e-2)
print(x)
y
```

```python
for step in range(1000):
    res = f(x, y) 
    err = (res - result_true)**2  

    err.backward()  
    optimizer.step() 
    optimizer.zero_grad()

    # Виводимо кожні 100 кроків
    if step % 100 == 0:
        print(f"Крок {step}: f(x,y)={res.item():.6f}, x={x.item():.4f}, y={y.item():.4f}")
```
